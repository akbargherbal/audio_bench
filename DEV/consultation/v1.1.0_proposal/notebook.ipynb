```python
import pandas as pd
import regex as re
import os
from pathlib import Path
import json
```

```python
with open('./data.json', encoding='utf-8') as f:
    data = json.loads(f.read())
```

```python
data['analysis']["fields"][0].keys()
```

```python
df = pd.DataFrame(data['analysis']).rename(columns={'fields': 'DATA'})
```

```python
df['label'] =  df['DATA'].apply(lambda x: x['label'])
df['field_name'] =  df['DATA'].apply(lambda x: x['field_name'])
df['files_to_review'] =  df['DATA'].apply(lambda x: x['files_to_review'])
df['skills_required'] =  df['DATA'].apply(lambda x: x['skills_required'])
df['questions'] =  df['DATA'].apply(lambda x: x['questions'])
```

```python
df['IN_PERSONA'] = df['DATA'].apply(lambda x: {k:v for k,v in x.items() if k not in "files_to_review questions".split()})
```

```python
skills = df['IN_PERSONA'].tolist()
for (idx, i) in enumerate(skills):
    with open(f'./{str(idx).zfill(2)}.txt', mode='w', encoding='utf-8') as f:
        f.write(json.dumps(i, ensure_ascii=False))
```

```python
df['IN_PERSONA'] = df['DATA'].apply(lambda x: {k:v for k,v in x.items() if k in "files_to_review questions".split()})

questions = df['IN_PERSONA'].tolist()
for (idx, i) in enumerate(questions):
    with open(f'./Q_{str(idx).zfill(2)}.txt', mode='w', encoding='utf-8') as f:
        f.write(json.dumps(i, ensure_ascii=False))
```
